# Heart Health: Exploratory Data Analysis

## Objective
Explore a public heart disease dataset using Python to understand
its structure, data quality, distributions, and selected associations
between clinical measurements and the recorded outcome.

## Methods
- Data inspection and quality checks
- Descriptive statistics
- Exploratory visualizations
- Group comparisons
- Limitations and reproducibility

## Important note
This is an educational analysis of a historical dataset, not a
diagnostic model or medical recommendation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
%pip -q install ucimlrepo
from ucimlrepo import fetch_ucirepo

heart_disease = fetch_ucirepo(id=45)

X = heart_disease.data.features
y = heart_disease.data.targets

df = pd.concat([X, y], axis=1)

# Rename the original target column to "target"
df = df.rename(columns={"num": "target"})

print("Column names:")
print(df.columns.tolist())

print("\nData types and non-missing counts:")
df.info()

print("\nFirst five records:")
display(df.head())

print("\nSummary statistics:")
display(df.describe().T)
```

```python
print("Missing values by column:")
display(df.isna().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnique values in each column:")
display(df.nunique().sort_values())
```

```python
print("Original outcome counts:")
display(df["target"].value_counts(dropna=False).sort_index())

# Create a binary outcome for exploratory analysis.
df["outcome_binary"] = (df["target"] > 0).astype(int)

df["outcome_group"] = df["outcome_binary"].map({
    0: "No recorded heart disease",
    1: "Recorded heart disease"
})

print("\nBinary outcome counts:")
display(df["outcome_group"].value_counts())

print("\nCheck for missing outcome labels:")
print(df["outcome_group"].isna().sum())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types and non-missing counts:")
df.info()

print("\nFirst five records:")
display(df.head())

print("\nSummary statistics:")
display(df.describe().T)
```

```python
print("Missing values by column:")
display(df.isna().sum().sort_values(ascending=False))

print("Duplicate rows:", df.duplicated().sum())

print("\nUnique values in each column:")
display(df.nunique().sort_values())
```

```python
print("Column names:", df.columns.tolist())

if "target" not in df.columns:
    raise ValueError(
        "Expected a 'target' column. Inspect the dataset columns "
        "and confirm the source before proceeding."
    )

print("\nOutcome counts:")
display(df["target"].value_counts(dropna=False).sort_index())

print("\nOutcome proportions:")
display(
    df["target"]
    .value_counts(normalize=True, dropna=False)
    .sort_index()
    .mul(100)
    .round(2)
    .rename("percentage")
)
```

```python
# Check that the required fields exist.
required_columns = [
    "age", "trestbps", "chol", "thalach", "target"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing expected columns: {missing_columns}")

# Check target values and basic numerical ranges.
print("Outcome values:", sorted(df["target"].dropna().unique()))
print("\nSelected variable ranges:")
display(df[required_columns].agg(["min", "max"]).T)

print("\nMissing values in analysis variables:")
display(df[required_columns].isna().sum())
```

```python
outcome_labels = {
    0: "No recorded heart disease",
    1: "Recorded heart disease"
}

df["outcome_group"] = df["target"].map(outcome_labels)

if df["outcome_group"].isna().any():
    raise ValueError("Unexpected outcome coding. Verify the source.")

display(df[["target", "outcome_group"]].head())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="outcome_group",
    hue="outcome_group",
    order=["No recorded heart disease", "Recorded heart disease"],
    palette="Set2",
    legend=False
)

plt.title("Distribution of Recorded Heart Disease Outcome")
plt.xlabel("Recorded outcome")
plt.ylabel("Number of records")
plt.xticks(rotation=10, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=15,
    kde=True
)

plt.title("Age Distribution in the Dataset")
plt.xlabel("Age (years)")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()

print("Mean age:", round(df["age"].mean(), 2))
print("Median age:", round(df["age"].median(), 2))
print("Age standard deviation:", round(df["age"].std(), 2))
```

```python
age_summary = (
    df.groupby("outcome_group", observed=True)["age"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(age_summary)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="outcome_group",
    y="age",
    hue="outcome_group",
    order=["No recorded heart disease", "Recorded heart disease"],
    palette="Set2",
    legend=False
)

plt.title("Age by Recorded Outcome")
plt.xlabel("Recorded outcome")
plt.ylabel("Age (years)")
plt.xticks(rotation=10, ha="right")
plt.tight_layout()
plt.show()
```

```python
bp_data = df.dropna(subset=["trestbps", "outcome_group"])

bp_summary = (
    bp_data.groupby("outcome_group", observed=True)["trestbps"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(bp_summary)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=bp_data,
    x="outcome_group",
    y="trestbps",
    hue="outcome_group",
    order=["No recorded heart disease", "Recorded heart disease"],
    palette="Set2",
    legend=False
)

plt.title("Resting Blood Pressure by Recorded Outcome")
plt.xlabel("Recorded outcome")
plt.ylabel("Resting blood pressure (mm Hg)")
plt.xticks(rotation=10, ha="right")
plt.tight_layout()
plt.show()
```

```python
hr_data = df.dropna(subset=["thalach", "outcome_group"])

hr_summary = (
    hr_data.groupby("outcome_group", observed=True)["thalach"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(hr_summary)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=hr_data,
    x="outcome_group",
    y="thalach",
    hue="outcome_group",
    order=["No recorded heart disease", "Recorded heart disease"],
    palette="Set2",
    legend=False
)

plt.title("Maximum Heart Rate by Recorded Outcome")
plt.xlabel("Recorded outcome")
plt.ylabel("Maximum heart rate achieved")
plt.xticks(rotation=10, ha="right")
plt.tight_layout()
plt.show()
```

```python
summary = (
    df.groupby("outcome_group", observed=True)
    .agg(
        records=("target", "size"),
        mean_age=("age", "mean"),
        mean_resting_bp=("trestbps", "mean"),
        mean_cholesterol=("chol", "mean"),
        mean_max_heart_rate=("thalach", "mean")
    )
    .round(2)
)

display(summary)
```

```python
# Save clean analysis outputs for reproducibility.
summary.to_csv("outcome_summary.csv")

age_summary.to_csv("age_summary.csv")
bp_summary.to_csv("resting_blood_pressure_summary.csv")
hr_summary.to_csv("maximum_heart_rate_summary.csv")

print("Summary files exported successfully.")
```
